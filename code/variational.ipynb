```python
def aggregate_rssi_data(df, lat_prec=5, lon_prec=5):
    """
    lat_prec/lon_prec: number of decimal places. 
    5 decimals is ~1.1 meters at the equator.
    4 decimals is ~11 meters.
    """
    
    # 1. Round coordinates to create spatial bins
    df_agg = df.copy()
    df_agg['lat_bin'] = df_agg['latitude'].round(lat_prec)
    df_agg['lon_bin'] = df_agg['longitude'].round(lon_prec)
    
    # 2. Aggressive GroupBy
    # We include 'ap' (Access Point) and 'floor' because RSSI 
    # is meaningless without knowing which AP you're talking to.
    group_cols = ['lat_bin', 'lon_bin', 'indoor', 'ap', 'floor']
    
    aggregated = df_agg.groupby(group_cols).agg(
        y_mean=('rssi_sample', 'mean'),
        y_std=('rssi_sample', 'std'),
        n_obs=('rssi_sample', 'count')
    ).reset_index()
    
    # 3. Sort by 'indoor' to satisfy the JAX group structure (Group 1 then Group 2)
    # Let's assume indoor=0 is Group 1, indoor=1 is Group 2
    # aggregated = aggregated.sort_values('indoor')
    aggregated.columns = ['latitude', 'longitude', 'indoor', 'ap', 'floor', 'rssi_sample', 'rssi_std', 'n_obs']
    
    return aggregated
```

```python
# sequoia_nonan = aggregate_rssi_data(sequoia[sequoia.rssi_signal_measured], lat_prec=4)
sequoia_nonan = sequoia[sequoia.rssi_signal_measured]

scaler = MinMaxScaler()
scaler.fit(pd.DataFrame(np.stack([TL_CORNER[::-1], BR_CORNER[::-1]]), columns=['longitude', 'latitude']))
# scaler.fit(sequoia_nonan[['longitude', 'latitude']])
X_train = pd.DataFrame()
X_train[['longitude', 'latitude']] = scaler.transform(sequoia_nonan[['longitude', 'latitude']])
X_train['indoor'] = sequoia_nonan['indoor'].astype(float).values

y_train = sequoia_nonan['rssi_sample']

X_train = jnp.array(X_train)

y_train = jnp.array(y_train)
y_mean, y_std = y_train.mean(), y_train.std()
# y_train = (y_train - y_mean) / y_std
```

```python
indoor_mask = X_train[:,2].astype(bool)
outdoor_mask = ~X_train[:,2].astype(bool)

stat = jnp.mean
stat1 = jnp.mean
stat2 = jnp.mean

in_mean = jnp.mean(y_train[indoor_mask])
out_mean = jnp.mean(y_train[indoor_mask])

# def m(obs):
#     return stat1(y_train[indoor_mask]) * obs[2] + stat2(y_train[outdoor_mask]) * (1-obs[2])

# def m(obs):
#     return stat(y_train)

def m(obs):
    return jnp.where(obs[2], in_mean, out_mean)
    
def dist(obs1, obs2):
    '''
    Location distance only
    '''
    return jnp.linalg.norm(obs1[:2] - obs2[:2], ord=2)**2


# def dist(obs1, obs2):
#     '''
#     Location + indoor/outdoor
#     '''
#     return jnp.linalg.norm(obs1 - obs2, ord=2)**2


# def dist(obs1, obs2):
#     '''
#     Location + weighted indoor/outdoor
#     '''
#     loc_dist = jnp.linalg.norm(obs1[:2] - obs2[:2], ord=2)**2
#     inout_dist = jnp.abs(obs1[2] - obs2[2])
#     return loc_dist + (loc_dist  / jnp.sqrt(1)) * inout_dist

scale = 1 #min([y_train[indoor_mask].var(), y_train[outdoor_mask].var()])
bandwidth = 1e-2
def K(obs1, obs2):
    d = dist(obs1, obs2)
    return scale * jnp.exp(-d / (2*bandwidth))

m_batch = jax.jit(jax.vmap(m))
K_batch = jax.jit(jax.vmap(jax.vmap(K, in_axes=(None, 0)), in_axes=(0, None)))

a_in, b_in, a_out, b_out = jnp.array([
    1+1e-3, 1e3,
    1+1e-3, 1e3
])
```

```python
indoor = X_train[:, 2].astype(bool)  # boolean mask, fixed
JITTER = 0

n = X_train.shape[0]
Kn = K_batch(X_train, X_train) + jnp.eye(n) * JITTER
dy = y_train - m_batch(X_train)

@jax.jit
def logprob(phi):
    '''
    Posterior log density of phi = (phi_in, phi_out) = (log var_in, log var_out)
    '''
    phi_in, phi_out = phi
    var_in, var_out = jnp.exp(phi)
    
    Ky = Kn.at[jnp.diag_indices(n)].add(jnp.where(indoor, var_in, var_out))
    L = jnp.linalg.cholesky(Ky, upper=False)
    
    gamma = jsp.linalg.solve_triangular(L, dy, lower=True)
    data_fit = -0.5 * jnp.sum(gamma**2)

    log_det = -jnp.sum(jnp.log(jnp.diag(L)))

    prior_in = -(a_in - 1) * phi_in - b_in / var_in
    prior_out = -(a_out - 1) * phi_out - b_out/ var_out

    return data_fit + log_det + prior_in + prior_out
```

```python
@jax.jit
def logprob(phi):
    '''
    Posterior log density of phi = (phi_in, phi_out) = (log var_in, log var_out)
    '''
    phi_in, phi_out = phi
    var_in, var_out = jnp.exp(phi)
    
    Ky = Kn.at[jnp.diag_indices(n)].add(jnp.where(indoor, var_in, var_out))
    
    Kyy, _ = jsp.sparse.linalg.cg(Ky, dy)
    data_fit = -0.5 * dy.T @ Kyy

    log_det = -0.5 * jnp.trace(jnp.log(Ky))

    prior_in = -(a_in - 1) * phi_in - b_in / var_in
    prior_out = -(a_out - 1) * phi_out - b_out/ var_out

    return data_fit + log_det + prior_in + prior_out
```

```python
k = 25
exp_phi0 = jnp.abs(jr.normal(jr.PRNGKey(305), shape=(k,2)) * 0.1 + y_train.var())
phi0 = jnp.log(exp_phi0)

solver = msvgd.MSVGD(logprob)
```

```python
start = time.time()
result = solver.solve(phi0, mitosis_splits=3, max_iter=20, atol=0.5)
end = time.time()
print(f"Time elapsed: {end - start:.4f}")
```
