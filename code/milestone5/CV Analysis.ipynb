```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
knn_cv = pd.read_csv('../gridsearch/knn_cv_mse_results.csv')
gibbs_cv = pd.read_csv('../milestone4/cv_gibbs_mse_results.csv')
```

```python
def plot_results(stype, method, filename=None):
    knn_mse = knn_cv[knn_cv['split_type'] == stype]['mse']
    gibbs_mse = gibbs_cv[gibbs_cv['split_type'] == stype]['mse']

    k = np.arange(knn_mse.shape[0])
    width = 0.35
    
    plt.bar(k-width/2, knn_mse, width, color=(255/255, 165/255, 0/255), label='KNN')
    plt.bar(k+width/2, gibbs_mse, width, color=(157/255, 193/255, 131/255), label='GP')
    plt.axhline(knn_mse.mean(), color=(205/255, 115/255, 0/255), ls='--')
    plt.axhline(gibbs_mse.mean(), color=(107/255, 143/255, 81/255), ls='--')

    # plt.ylim(min([knn_mse.mean(), gibbs_mse.mean()]) / 1.5)
    plt.xlabel('Split Index'); plt.ylabel('Validation MSEs')
    plt.title(f'{k.shape[0]}-fold Validation MSE for {method} Splitting')

    plt.legend()
    if filename:
        plt.savefig(filename, dpi=400)
    plt.show()
```

```python
plot_results('rand', 'Random', 'random.png')
```

```python
plot_results('geo', 'Geographic', 'geographic.png')
```
